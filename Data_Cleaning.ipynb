```python
# Import required libraries

import pandas as pd
import numpy as np
```

```python
# Load the dataset

df = pd.read_csv(
    r"d:\My Projects\Sample - Superstore.csv",
    encoding="latin-1"
)


# Display first few rows of the dataset

print(df.head())
```

```python
# Display dataset structure and data types

print(df.info())
```

```python
# Generate descriptive statistics for numerical and categorical columns

print(df.describe(include='all'))
```

```python
# Standardize column names by removing extra spaces

df.columns = df.columns.str.strip()
```

```python
# Convert date columns to datetime format

df["Order Date"] = pd.to_datetime(df["Order Date"], errors='coerce')
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors='coerce')


# Verify updated data types

print(df.dtypes)
```

```python

# Check missing values in each column

print(df.isnull().sum())
```

```python
# Fill missing values in categorical columns

categorical_cols = ['Segment', 'State', 'Region', 'Sub-Category']

for col in categorical_cols:
    df[col] = df[col].fillna('Unknown')
```

```python

# Fill missing values in numerical columns using median

numerical_cols = ['Quantity', 'Discount', 'Profit']

for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python

# Verify missing values after treatment

print(df.isnull().sum())
```

```python

# Check duplicate records

print(df.duplicated().sum())
```

```python
# Remove duplicate rows

df.drop_duplicates(inplace=True)

# Verify duplicate removal

print(df.duplicated().sum())
```

```python
# Clean and standardize text columns

string_cols = df.select_dtypes(include='object').columns

for col in string_cols:
    df[col] = (
        df[col]
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
    )
```

```python
# Standardize text formatting using title case

title_cols = [
    'Ship Mode',
    'Customer Name',
    'Segment',
    'City',
    'State',
    'Region',
    'Category',
    'Sub-Category',
    'Product Name'
]

for col in title_cols:
    if col in df.columns:
        df[col] = df[col].str.title()
```

```python

# Drop unnecessary columns

df.drop(columns='Country', inplace=True)
```

```python
# Verify cleaned dataset

print(df.head())
```

```python
# Export cleaned dataset

df.to_csv("cleaned_superstore.csv", index=False)
```
